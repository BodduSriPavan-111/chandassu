### Visualize Scores

```python
import matplotlib.pyplot as plt

import sys
import os

sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))

from chandassu.telugu.laghuvu_guruvu import LaghuvuGuruvu
from chandassu.telugu.padya_bhedam import check_padyam
```

```python
all_types= ["kandamu","aataveladi","teytageethi","seesamu","vutpalamaala","champakamaala","saardulamu","mattebhamu"]
```

```python
padyam= """శ్రీరఘురామ చారుతులసీదళదామ శమక్షమాది శృం
గారగుణాభిరామ త్రిజగన్నుతశౌర్యరమాలలామ దు
ర్వార కబంధరాక్షసవిరామ జగజ్జనకల్మషార్ణవో
త్తారకనామ భద్రగిరి దాశరథీ! కరుణాపయోనిధీ"""

lg_data= LaghuvuGuruvu(data= padyam).generate()         
type_score= []
for i in all_types:
    score= check_padyam(
                        lg_data= lg_data ,
                        type= i,
                        return_micro_score= True, 
                        verbose= False
                    )
    print(i,score)
    type_score.append([score['chandassu_score'],i])
```

```python
scores=  [i[0] for i in type_score]
padyam_type = [i[1] for i in type_score]
```

```python
plt.figure(figsize= (13,6))
plt.bar(range(len(scores)),scores,color= "cornflowerblue",width=0.4)
plt.xticks(range(len(scores)),padyam_type)
plt.show()
```
