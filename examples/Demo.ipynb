### Import

Toṇḍamu nēka dantamunu tōrapu bojjayu vāma hastamun </br>
meṇḍuga mrōyu gajjelunu mellani cūpulu mandahāsamun </br>

```python
data= """తొండము నేక దంతమును తోరపు బొజ్జయు వామ హస్తమున్
మెండుగ మ్రోయు గజ్జెలును మెల్లని చూపులు మందహాసమున్
కొండొక గుజ్జు రూపమున కోరిన విద్యలకెల్ల నొజ్జవై
యుండెడి పార్వతీ తనయ యోయి గణాధిప నీకు మ్రొక్కెదన్"""
```

```python
lg= LaghuvuGuruvu( data= data )
```

### Aksharam Tokenization (Character-level Prosody aware tokenization)

```python
import regex as re
print( re.findall(r"\X", data) )
```

```python
tokens= lg.tokenize()
print( tokens )
```

### Syllabic Weight Classification
#### 1. Laghuvu: Short pronounced ( '|' )
#### 2. Guruvu: Long pronounced ( 'U' )

```python
lg_data= lg.generate()
print( lg_data )
```

### Scoring

- n_paadalu: No.of lines
- gana_kramam: Syllabic sequence
- yati_sthanam: Caesura Position
- N Aksharalu: N Character Tokens
- Prasa: Alliteration

```python
score= check_padyam( 
                        lg_data= lg_data ,
                        type= "vutpalamaala",
                        return_micro_score= True, 
                        verbose= False
                    )
print( score )
```

```python
score= check_padyam( 
                        lg_data= lg_data ,
                        type= "champakamaala",
                        return_micro_score= True, 
                        verbose= False
                    )
print( score )
```

```python
score= check_padyam( 
                        lg_data= lg_data ,
                        type= "kandamu",
                        return_micro_score= True, 
                        verbose= False
                    )
print( score )
```

### Define your own Padyam (Poem)

```python
from chandassu.telugu.customization import Custom
from chandassu.telugu.ganam import *
```

```python
#Reference: https://te.wikipedia.org/wiki/%E0%B0%AE%E0%B0%A4%E0%B1%8D%E0%B0%A4%E0%B0%95%E0%B1%8B%E0%B0%95%E0%B0%BF%E0%B0%B2

custom= Custom( 
type= "mattakokila", 
n_paadalu= 4, 
n_aksharalu= 20, 
gana_kramam= ( ( ra_ganam, sa_ganam, ja_ganam, ja_ganam, bha_ganam, ra_ganam),   
                ( ra_ganam, sa_ganam, ja_ganam, ja_ganam, bha_ganam, ra_ganam),
                ( ra_ganam, sa_ganam, ja_ganam, ja_ganam, bha_ganam, ra_ganam),
                ( ra_ganam, sa_ganam, ja_ganam, ja_ganam, bha_ganam, ra_ganam) ),
yati_sthanam= (4,1), 
yati_paadalu= (1,2,3,4), 
prasa= True, 
only_generic_yati= True
)
```

```python
data= """రాజభూషణ నిత్యసత్య సరస్వతీవిలసన్ముఖాం
భోజ రాజమనోజ భూజనపూజ్యమాన మహాయశో
రాజహంస పయోజినీవనరమ్య దిఙ్ముఖ విక్రమో
ద్వేజితాహిత విష్ణుసన్నిభ విష్ణువర్ధనభూపతీ"""
```

```python
lg_data= LaghuvuGuruvu( data= data ).generate()
```

```python
custom.check_padyam( lg_data= lg_data, verbose= False )
```
